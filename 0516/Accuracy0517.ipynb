```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Load history
# =========================
history = np.load("training_history_gaf_cnn_vit_mlp.npz")

train_acc = history["train_acc"]
val_acc = history["val_acc"]

epochs = np.arange(1, len(train_acc) + 1)

# =========================
# Plot
# =========================
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(
    epochs, train_acc,
    color="blue",
    linewidth=2.2,
    label="Train"
)

ax.plot(
    epochs, val_acc,
    color="red",
    linewidth=2.2,
    label="Val"
)

ax.set_xlabel("Epoch", fontsize=14)
ax.set_ylabel("Accuracy", fontsize=14)
# ax.set_title("Training and Validation Accuracy Curves", fontsize=15)

ax.tick_params(axis='both', labelsize=12)
ax.legend(fontsize=12, frameon=True)
# ax.grid(True, linestyle='--', linewidth=0.6, alpha=0.6)

# 去掉上边框和右边框，更适合论文风格
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()

# =========================
# Save high-quality figures
# =========================
fig.savefig("accuracy_curve.png", dpi=600, bbox_inches="tight")
# fig.savefig("accuracy_curve.pdf", bbox_inches="tight")
# fig.savefig("accuracy_curve.tiff", dpi=600, bbox_inches="tight")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Load history
# =========================
history = np.load("training_history_gaf_cnn_vit_mlp.npz")

train_acc = history["train_acc"]
val_acc = history["val_acc"]

# Convert accuracy to error rate
train_error = 1 - train_acc
val_error = 1 - val_acc

epochs = np.arange(1, len(train_error) + 1)

# =========================
# Plot
# =========================
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(
    epochs, train_error,
    color="blue",
    linewidth=2.2,
    label="Train"
)

ax.plot(
    epochs, val_error,
    color="red",
    linewidth=2.2,
    label="Val"
)

ax.set_xlabel("Epoch", fontsize=14)
ax.set_ylabel("Error rate", fontsize=14)

# y-axis range
ax.set_ylim(0, 0.5)
ax.set_yticks(np.arange(0, 0.405, 0.2))

ax.tick_params(axis='both', labelsize=12)
ax.legend(fontsize=12, frameon=True)

# Remove top and right borders
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()

# =========================
# Save high-quality figure
# =========================
fig.savefig("error_curve.png", dpi=600, bbox_inches="tight")

plt.show()
```
