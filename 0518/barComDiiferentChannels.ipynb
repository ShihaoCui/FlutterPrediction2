```python
import matplotlib.pyplot as plt
import numpy as np

# Accuracy results
channels = ['Ch1-7', 'Ch1-3', 'Ch4', 'Ch5-7', 'Ch5', 'Ch6', 'Ch7']
acc = np.array([0.956, 0.8041, 0.8364, 0.9583, 0.9156, 0.9187, 0.94276])

# Convert to percentage
acc_percent = acc * 100

# Different colors for bars
colors = [
    '#4C72B0',  # Ch1-7
    '#55A868',  # Ch1-3
    '#C44E52',  # Ch4
    '#8172B3',  # Ch5-7
    '#CCB974',  # Ch5
    '#64B5CD',  # Ch6
    '#DD8452'   # Ch7
]

# Plot
plt.figure(figsize=(8, 5))
bars = plt.bar(channels, acc_percent, width=0.65, color=colors)

# Axis labels
plt.xlabel('Input channels', fontsize=13)
plt.ylabel('Accuracy (%)', fontsize=13)

# Axis settings
plt.ylim(75, 100)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Tick font
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# Add value labels
for bar, value in zip(bars, acc_percent):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.5,
        f'{value:.2f}%',
        ha='center',
        va='bottom',
        fontsize=11
    )

plt.tight_layout()

# Save figure
plt.savefig('Accuracy_Different_Channels.png', dpi=600, bbox_inches='tight')
plt.savefig('Accuracy_Different_Channels.pdf', bbox_inches='tight')

plt.show()
```
